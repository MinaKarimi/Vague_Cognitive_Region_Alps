```python
# Function to compute Jaccard between two boolean lists
def jaccard(a, b):
    intersection = sum(x and y for x, y in zip(a, b))
    union = sum(x or y for x, y in zip(a, b))
    return intersection / union if union else 0
```

```python
def dice(a, b):
    intersection = sum(x and y for x, y in zip(a, b))
    count_a = sum(a)
    count_b = sum(b)
    denom = count_a + count_b
    return (2 * intersection / denom) if denom else 0
```

```python
def confusion_matrix_indices(matrix, labels, cmap, matrix_label, title, output_file):
    plt.figure(figsize=(10, 8))
    plt.imshow(matrix, vmin=0, vmax=1, cmap=cmap)
    
    plt.colorbar(label=matrix_label)
    
    plt.xticks(ticks=np.arange(len(labels)), labels=labels, rotation=45, ha='right')
    plt.yticks(ticks=np.arange(len(labels)), labels=labels)

    # Add text annotations
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            plt.text(j, i, f"{matrix[i, j]:.2f}", ha="center", va="center")
    
    plt.title(title)
    plt.tight_layout()
    plt.savefig(output_file, bbox_inches="tight")
    plt.show()
```

```python
def plot_model_confusion_matrix(df, model, metric_name):
    """
    Plot a confusion matrix heatmap from a DataFrame.

    Args:
        df (pd.DataFrame): DataFrame where rows are source languages,
                           columns are target languages or models.
        model (str): Name of the model to select the column or slice.
        metric_name (str): Metric name for the title (e.g., "Jaccard").
    """
    # If the DataFrame has multi-index or columns for models
    if model not in df.columns:
        raise ValueError(f"Model '{model}' not found in DataFrame columns.")
    
    # Extract the matrix for this model
    # Assuming df is square: rows=source languages, columns=target languages, values=model metric
    data = df[model]
    
    # If 'data' is a Series, convert to DataFrame
    if isinstance(data, pd.Series):
        data = pd.DataFrame(data).T  # transpose to make it 2D
    
    # Plot heatmap
    plt.figure(figsize=(7,6))
    sns.heatmap(data, annot=True, fmt=".3f", vmin=0, vmax=1, cmap="Blues",
                xticklabels=data.columns, yticklabels=data.index, square=True)
    plt.title(f"{metric_name} Confusion Matrix for {model}")
    plt.xlabel("Target Language")
    plt.ylabel("Source Language")
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

def plot_model_confusion_matrix(df, model, metric_name, cmap):
    df = df.copy()
    df.index = df.index.astype(str)
    df.columns = df.columns.astype(str)

    labels = [x for x in df.index if f"_{model}_" in x]

    if not labels:
        raise ValueError(
            f"No rows found for model='{model}'. "
            f"Available labels: {df.index.tolist()}"
        )

    matrix = df.loc[labels, labels]

    # Extract language codes
    matrix.index = [x.split("_")[-1] for x in matrix.index]
    matrix.columns = [x.split("_")[-1] for x in matrix.columns]

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        matrix,
        annot=True,
        fmt=".2f",
        cmap=cmap,
        square=True,
        vmin=0,
        vmax=1,
        cbar_kws={"label": metric_name}
    )

    plt.title(f"{metric_name} – {model}")
    plt.xlabel("Target Language")
    plt.ylabel("Source Language")

    filename = f"{model}_{metric_name}.pdf"
    plt.savefig(filename, bbox_inches="tight")
    plt.show()

    print(f"Saved: {filename}")
```

```python
languages = ["En", "Ge", "Fr", "It", "Sl"]
```

```python
model_pairs = {
    "GPT4o–DS":  ("GPT4o_", "DS_"),
    "GPT4o–GMN": ("GPT4o_", "GMN_"),
    "DS–GMN":    ("DS_",    "GMN_"),
    "GPT55–GPT4o":    ("GPT55_",    "GPT4o_"),
    "GPT55–DS":    ("GPT55_",    "DS_"),
    "GPT55–GMN":    ("GPT55_",    "GMN_"),
}
```

```python
def extract_metric_by_language(
    df_metric,
    field_prefix="B_",   # "B_" for Jaccard/Dice, "S_" for Pearson
):
    out = {k: [] for k in model_pairs}

    for lang in languages:
        for label, (m1, m2) in model_pairs.items():
            f1 = f"{field_prefix}{m1}{lang}"
            f2 = f"{field_prefix}{m2}{lang}"

            out[label].append(df_metric.loc[f1, f2])

    return out
```

```python
def plot_metric_by_language(
    metric_by_language,
    title,
    ylabel,
    save_name,
    save_dir="plots",
):
    x = np.arange(len(languages))
    n_series = len(metric_by_language)
    width = 0.8 / n_series

    fig, ax = plt.subplots(figsize=(10, 5.5))

    # Center each group of bars around its language tick.
    for i, (label, values) in enumerate(metric_by_language.items()):
        offset = (i - (n_series - 1) / 2) * width

        ax.bar(
            x + offset,
            values,
            width,
            label=label,
            color=colors[i % len(colors)]
        )

    ax.set_xticks(x)
    ax.set_xticklabels(languages)
    ax.set_xlabel("Language")
    ax.set_ylabel(ylabel)
    ax.set_title(title, pad=12)

    if ylim is not None:
        ax.set_ylim(*ylim)

    # Subtle horizontal grid lines behind the bars.
    ax.set_axisbelow(True)
    ax.grid(axis="y", color="lightgray", linewidth=0.6, alpha=0.5)

    # Horizontal legend outside the plotting area.
    ax.legend(
        loc="upper center",
        bbox_to_anchor=(0.5, -0.18),
        ncols=n_series,
        frameon=False,
        fontsize=9,
        columnspacing=1.2,
        handlelength=1.5
    )

    fig.tight_layout()

    os.makedirs(save_dir, exist_ok=True)
    filename = f"{save_name}.pdf"
    save_path = os.path.join(save_dir, filename)

    fig.savefig(save_path, format="pdf", bbox_inches="tight")
    plt.show()

    print(f"Saved: {save_path}")
```
