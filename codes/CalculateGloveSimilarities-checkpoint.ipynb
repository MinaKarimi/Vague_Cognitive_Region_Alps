```python
def extrat_authorities(layer_name):
    
    layer = next((lyr for lyr in active_map.listLayers() if lyr.name == layer_name), None)

    if layer:
        # Dynamically list fields for "country" and all "Name_*" fields
        all_fields = [field.name for field in arcpy.ListFields(layer.dataSource)]
        selected_fields = [field for field in all_fields if field == "COUNTRY" or field.startswith("NAME_")]
        print("Selected Fields:", selected_fields)  # Optional: Print the selected fields

        # Include "OID@" for feature IDs and "SHAPE@" for geometry if needed
        fields = ["OID@"] + selected_fields
        # Create an empty list to store feature data
        data = []

        # Use SearchCursor to read data from the layer
        with arcpy.da.SearchCursor(layer.dataSource, fields) as cursor:
            for row in cursor:
                data.append(row)

        # Create a DataFrame from the data
        df = pd.DataFrame(data, columns=fields)
        # Optional: Display the DataFrame
        #print(df)
        return df

    else:
        print(f"Layer '{layer_name}' not found.")
        
```

```python
# Load GloVe embeddings into a dictionary

def load_embeddings(file_path):
    embeddings = {}
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            try:
                values = line.split()
                #print(values[1:])
                word = values[0]
                vector = np.asarray(values[1:], dtype='float32')
                embeddings[word] = vector

            except:
                pass

    return embeddings

glove_embeddings_path = 'C:/Users/Mina/Desktop/Mina Files/PostDoc Vienna/Codes/glove.840B.300d/glove.840B.300d.txt'  # Adjust the path to your downloaded GloVe file
glove_embeddings = load_embeddings(glove_embeddings_path)
```

```python
def get_vector(word, glove_embeddings):
    """Retrieve the GloVe vector for a word, or return a zero vector if not found."""
    return glove_embeddings.get(word, np.zeros(300))  # Adjust the dimensionality if different
```

```python
def get_combined_vector(multipart_word, glove_embeddings):
    """
    Handle multipart words by averaging the vectors of individual parts.
    If all parts are missing, return a zero vector.
    """
    words = multipart_word.split()  # Split by space
    vectors = [get_vector(word, glove_embeddings) for word in words if word in glove_embeddings]
    if vectors:
        return np.mean(vectors, axis=0)  # Average the vectors
    return np.zeros(300)  # Return a zero vector if no parts are found
```

```python
def calculate_words_similarity(word1, word2, glove_embeddings):
    """
    Calculate similarity between two words, handling multipart words.
    """
    # Check if word1 or word2 is multipart and get combined vectors
    vec1 = get_combined_vector(word1, glove_embeddings) if " " in word1 else get_vector(word1, glove_embeddings)
    vec2 = get_combined_vector(word2, glove_embeddings) if " " in word2 else get_vector(word2, glove_embeddings)

    # Handle zero vectors (if both words are OOV)
    if not np.any(vec1) or not np.any(vec2):  # Check if vectors are zero
        return 0.0  # Return similarity of 0.0 for OOV cases

    # Calculate cosine similarity
    similarity = 1 - cosine(vec1, vec2)
    return similarity
```

```python
def cosine_similarity(vec1, vec2):
    norm1 = np.linalg.norm(vec1)
    norm2 = np.linalg.norm(vec2)
    if norm1 == 0 or norm2 == 0:  # Handle zero vectors
        return 0.0
    return np.dot(vec1, vec2) / (norm1 * norm2)
```

```python
#Function to calculate similarities for an array of strings and a target word
def calculate_similarities(strings_array, target_word, glove_embeddings):
    # Ensure input is a list
    if not isinstance(strings_array, (list, pd.Series)):
        raise ValueError("Input 'strings_array' must be a list or pandas Series of strings.")

    # Compute similarities
    similarities = [calculate_words_similarity(string, target_word, glove_embeddings) for string in strings_array]

    # Create a DataFrame
    results_df = pd.DataFrame({
        "ObjectID": range(1, len(strings_array) + 1),
        "String": strings_array,
        "Similarity": similarities
    })

    return results_df
```
